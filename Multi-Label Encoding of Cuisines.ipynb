```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('zomato.csv')
```

```python
df.head()
```

```python
df['cuisines'].head(10)
```

```python
df['cuisines'].isnull().sum()
```

```python
df['cuisines'] = df['cuisines'].fillna('Unknown')
```

```python
df['cuisines'] = df['cuisines'].str.strip()
```

```python
df['cuisines_list'] = df['cuisines'].apply(
    lambda x: [i.strip() for i in x.split(',')]
)
```

```python
df[['cuisines', 'cuisines_list']].head(10)
```

```python
all_cuisines = set()

for cuisines in df['cuisines_list']:
    all_cuisines.update(cuisines)

print("Total Unique Cuisines:", len(all_cuisines))
```

```python
sorted(all_cuisines)
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

mlb = MultiLabelBinarizer()

cuisine_encoded = mlb.fit_transform(df['cuisines_list'])
```

```python
cuisine_df = pd.DataFrame(
    cuisine_encoded,
    columns=mlb.classes_,
    index=df.index
)
```

```python
cuisine_df.head()
```

```python
df = pd.concat([df, cuisine_df], axis=1)
```

```python
df.head()
```

```python
print("Total Cuisine Columns:", len(mlb.classes_))
```

```python
cuisine_df.sum().sort_values(ascending=False).head(10)
```

```python
df.to_csv('zomato_cuisine_encoded.csv', index=False)
```
